```python
import matplotlib.pyplot as plt
import pandas as pd

import pyam
import ixmp4
```

```python
platform = ixmp4.Platform("scenariomip-cmip7")
```

```python
df = pyam.read_ixmp4(platform, region="World", variable="Emissions|Kyoto Gases")
```

```python
df
```

```python
scenario_color_mapping = {
    "High": "#A41212",
    "High-to-Low": "#E744F6",
    "Medium": "#FC7B03",
    "Medium-to-Low": "#dec820",
    "Low": "#20A359",
    "Low-to-Negative": "#22e5db",
    "Very Low": "#16188F",
}
```

```python
df.scenario
```

```python
df
```

```python
ensemble_args = dict(model="Ensemble", region="World", variable="Emissions|Kyoto Gases", unit="Mt CO2-equiv/year")

ensemble_list = list()

def make_df(x, scenario, limit):
    return (
        pyam.IamDataFrame(
            x.reset_index(name="value").rename(columns={"index": "year"}),
            scenario=f"{scenario} [{limit}]",
            **ensemble_args,
        )
    )

for scenario, color in scenario_color_mapping.items():
    data = df.filter(scenario=f"{scenario} *", year=range(2020, 2101, 5)).interpolate(range(2025, 2096, 5)).timeseries()
    ensemble_list.append(make_df(data.max(), scenario, "Max"))
    ensemble_list.append(make_df(data.min(), scenario, "Min"))
```

```python
ensemble = pyam.concat(ensemble_list)
```

```python
ensemble
```

```python
fig, ax = plt.subplots()

for scenario, color in scenario_color_mapping.items():
    ensemble.filter(scenario=f"{scenario} *").plot(ax=ax, color=color, lw=1, linestyle="--", legend=False)
    df.filter(scenario=f"{scenario} *(Marker)").plot(ax=ax, color=color, lw=2, label=scenario, legend=True)

ax.set_title(None)
```

```python
ensemble.filter(scenario=f"Very Low *").plot(ax=ax, color=color, lw=1, linestyle="--")
```

```python

for scenario, color in scenario_color_mapping.items():
    df.filter(scenario=f"{scenario}*").plot(ax=ax, color=color, fill_between=True)
```
